14. Чем PyPy отличается от CPython?

PyPy — альтернативная реализация Python с JIT-компилятором, на долгих числовых циклах обычно заметно быстрее CPython (в моём тесте примерно в 4 раза)

15. Почему PyPy не может использоваться для всех проектов на Python?

Неполная совместимость с расширениями на C (например, часть научных библиотек), больший расход памяти, выигрыш только на долгих программах (на коротких JIT не успевает «разогреться»)
